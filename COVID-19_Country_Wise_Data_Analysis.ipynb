```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the Dataset

df = pd.read_csv("country_wise_latest.csv") 
df.head()
```

```python
# Display Dataset Information
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
# Check Missing Values
df.isnull().sum()
```

```python
# Check Duplicate Records
df.duplicated().sum()

# Remove duplicates if they exist:

df = df.drop_duplicates()
df.shape
```

```python
# Basic Statistical Analysis
df.describe()
```

```python
# Check WHO Regions
df["WHO Region"].value_counts()
```

```python
# Top 10 Countries by Confirmed Cases
top_confirmed = df.sort_values( by="Confirmed", ascending=False ).head(10)   
# Visualization
plt.figure(figsize=(12, 6)) 
plt.bar( top_confirmed["Country/Region"], top_confirmed["Confirmed"], color='green')
plt.title("Top 10 Countries by Confirmed COVID-19 Cases") 
plt.xlabel("Country") 
plt.ylabel("Confirmed Cases") 
plt.xticks(rotation=45) 
plt.tight_layout()
```

```python
# Top 10 Countries by Deaths
top_deaths = df.sort_values( by="Deaths", ascending=False ).head(10) 
# Visualization
plt.figure(figsize=(12, 6)) 
sns.barplot( data=top_deaths, x="Country/Region", y="Deaths",color='green') 
plt.title("Top 10 Countries by COVID-19 Deaths") 
plt.xlabel("Country")
plt.ylabel("Deaths")
plt.xticks(rotation=45)
plt.tight_layout() 
plt.show()
```

```python
# Top 10 Countries by Active Cases
top_active = df.sort_values( by="Active", ascending=False ).head(10)
# Visualization
plt.figure(figsize=(12, 6)) 
plt.bar( top_active["Country/Region"], top_active["Active"], color='green')
plt.title("Top 10 Countries by Active COVID-19 Cases") 
plt.xlabel("Country") 
plt.ylabel("Active Cases") 
plt.xticks(rotation=45) 
plt.tight_layout() 
plt.show()
```

```python
# Calculate total cases for each WHO region:
region_cases = df.groupby("WHO Region")[["Confirmed", "Deaths", "Recovered", "Active"]].sum()
region_cases
```

```python
# Confirmed Cases by WHO Region
plt.figure(figsize=(10, 6))
region_cases["Confirmed"].sort_values(ascending=False).plot(kind="bar", color='green')
plt.title("Confirmed COVID-19 Cases by WHO Region")
plt.xlabel("WHO Region")
plt.ylabel("Confirmed Cases")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Correlation Analysis
numeric_columns = ["Confirmed","Deaths","Recovered","Active","New cases","New deaths","New recovered","Confirmed last week","1 week change","1 week % increase"]
correlation = df[numeric_columns].corr()
correlation
```

```python
# Correlation Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(correlation,annot=True ,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap of COVID-19 Dataset")
plt.tight_layout()
plt.show()
```

```python
# Confirmed Cases vs Deaths
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df,x="Confirmed",y="Deaths",color='green',alpha = 1)
plt.title("Relationship Between Confirmed Cases and Deaths")
plt.xlabel("Confirmed Cases")
plt.ylabel("Deaths")
plt.tight_layout()
plt.show()
```

```python
# Confirmed Cases vs Recovered Cases
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df,x="Confirmed",y="Recovered",color="green",alpha=1)
plt.title("Relationship Between Confirmed and Recovered Cases")
plt.xlabel("Confirmed Cases")
plt.ylabel("Recovered Cases")
plt.tight_layout()
plt.show()
```

```python
# Distribution of Confirmed Cases
plt.figure(figsize=(10, 6))
sns.histplot(df["Confirmed"],bins=30,kde=True,color="green")
plt.title("Distribution of Confirmed COVID-19 Cases")
plt.xlabel("Confirmed Cases")
plt.ylabel("Number of Countries")
plt.tight_layout()
plt.show()
```

```python
# Distribution of Deaths
plt.figure(figsize=(10, 6))
sns.histplot(df["Deaths"],bins=30,kde=True,color="green")
plt.title("Distribution of COVID-19 Deaths")
plt.xlabel("Deaths")
plt.ylabel("Number of Countries")
plt.tight_layout()
plt.show()
```

```python
# Overall COVID-19 Statistics
total_confirmed = df["Confirmed"].sum()
total_deaths = df["Deaths"].sum()
total_recovered = df["Recovered"].sum()
total_active = df["Active"].sum()
overall_statistics = pd.DataFrame({"Statistic": ["Total Confirmed Cases","Total Deaths","Total Recovered","Total Active Cases"],
"Value": [total_confirmed,total_deaths,total_recovered,total_active]})
overall_statistics
```

```python
# Overall Recovery and Death Rate
# Overall Recovery and Death Rate
overall_recovery_rate = (total_recovered / total_confirmed) * 100
overall_death_rate = (total_deaths / total_confirmed) * 100
rates = pd.DataFrame({"Rate": ["Overall Recovery Rate","Overall Death Rate"],
"Percentage": [round(overall_recovery_rate, 2),round(overall_death_rate, 2)]})
rates
```

```python
# Save Cleaned Dataset
df.to_csv("cleaned_country_wise_latest.csv",index=False)
```
